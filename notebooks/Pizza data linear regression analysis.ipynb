```python
# importing required libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split 
```

```python
# Data
pizza_size= np.array([[6],[8],[10],[12],[14],[16]])
pizza_price=np.array([130, 155, 210, 235, 290, 310])
```

```python
# Creating Dataframe
df = pd.DataFrame({
    "pizza_size": pizza_size.flatten(),
    "pizza_price":pizza_price
})
df
```

```python
# Saving dataframe to csv file
df.to_csv("Pizza_data.csv", index=False)
```

```python
df = pd.read_csv("Pizza_data.csv")
df
```

```python
# Create Model
model=LinearRegression()
```

```python
# Train Model
model.fit(pizza_size,pizza_price)
```

```python
# Prediction
y_pred=model.predict(pizza_size)
print("Predicted Price:",y_pred)
```

```python
# Visualization
plt.scatter(pizza_size,pizza_price,color='blue',label='original')
plt.plot(pizza_size,y_pred,color='red',label='predicted')
plt.title=('Linear Regression Pizza data sample')
plt.xlabel=('Pizza Size (inches)')
plt.ylabel=('Pizza Price (₹)')
plt.legend()
plt.show()
```

```python
new_pizza = [[18],[20]]                                   # Input: 18-20 inch pizza
predicted_price = model.predict(new_pizza)                # Model predicts
print("18 inch Pizza Price:", int(predicted_price[0]))    # Show results
print("20 inch Pizza Price:", round(predicted_price[1]))
```
